Purpose of the notbook is to look at the mysterious contrast boundary that showed up in most projection from the andor 3 zyla. 
For this purpose that datset of 6 projections from 077_calibration is imported. It shows darkfield images and has 2 projections in which the boundary doesn't show, 
as well as 4 in which it does. Maybe this can give an insight into why and when it happens, what the root cause is and  how to best mitigate it. 

```python
import h5py
import numpy as np
import matplotlib as plt
```

```python
from pathlib import Path

data_dir = Path(
    "/asap3/spec.instruments/gpfs/demonstrator/2026/data/"
    "11024317/raw/077_calibration/zyla"
)

def load_scan(scan_number):
    files = sorted(
        path for path in data_dir.glob(f"scan_{scan_number}_*")
        if path.is_file() and path.suffix in {".nx", ".h5"}
    )
    frames = []
    for path in files:
        with h5py.File(path, "r") as f:
            data = f["/entry/instrument/detector/data"][:]
        # Normalize a single image to (1, height, width).
        if data.ndim == 2:
            data = data[np.newaxis, ...]
        if data.ndim != 3:
            raise ValueError(f"Unexpected shape {data.shape} in {path.name}")
        frames.append(data)
    return np.concatenate(frames, axis=0)


darks_arr = load_scan(247)import numpy as np
#flats_arr = load_scan(240)

print("Darks:", darks_arr.shape, darks_arr.dtype,
      "min/max:", darks_arr.min(), darks_arr.max())
#print("Flats:", flats_arr.shape, flats_arr.dtype,
#      "min/max:", flats_arr.min(), flats_arr.max())
```
